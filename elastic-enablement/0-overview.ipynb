## Elastic in 30 Seconds

**Elastic** (NYSE: `ESTC`) calls itself *"the Search AI Company."* It builds **Elasticsearch**, a fast engine for storing and searching large volumes of data, and sells three solutions on top of it:

1. **Search & AI**
2. **Observability**
3. **Security**

> Elastic helps organizations find answers in all their data in real time, whether that's a shopper searching a catalog, an engineer finding why an app crashed, or a security analyst hunting an attacker. Today it is also the data layer that grounds AI agents in a company's own information.

**Elastic vs. Elasticsearch:** Elasticsearch is the product (the engine). Elastic is the company, which was called Elasticsearch until it rebranded in 2015.

## The Elastic Stack

The **Elastic Stack** (formerly the "ELK Stack") moves data from where it is created to where people use it.

```
Elastic Agent / Beats  -->  Logstash  -->  Elasticsearch  -->  Kibana
 (collect data)           (transform)     (store + search)    (visualize)
```

| Component | What it does | Analogy |
| --- | --- | --- |
| **Elastic Agent / Beats** | Lightweight collectors on servers, laptops, and cloud apps | Delivery trucks picking up logs, metrics, and events |
| **Logstash** | Optional pipeline that cleans and transforms data | The mailroom that sorts and labels |
| **Elasticsearch** | Distributed search and analytics engine and data store | The library with an instant card catalog |
| **Kibana** | Web UI for search, dashboards, alerts, and admin | The reading room with charts |

## Key Terms

| Term | Meaning |
| --- | --- |
| **Index** | A collection of similar documents (roughly a "table") |
| **Document** | One record stored as JSON (roughly a "row") |
| **Inverted index** | Maps each word to the documents containing it, like the index at the back of a book; this is why search is fast |
| **Cluster / Node / Shard** | Data is split into shards across many machines (nodes) in a cluster, so it scales and survives failures |
| **Relevance** | Ranking results by how well they match the query (classic algorithm: BM25) |
| **Full-text vs. vector search** | Matching words vs. matching meaning; *hybrid search* combines both |
| **ES\|QL** | Elastic's piped query language for searching and analyzing data |
| **Logs, metrics, traces** | The three main types of machine data an application produces |
| **OpenTelemetry (OTel)** | Open standard for collecting machine data; Elastic supports it natively |
| **SIEM** | Security Information and Event Management: collecting security logs and detecting threats |
| **RAG** | Retrieval-Augmented Generation: retrieve relevant company data first, then let an LLM answer |

## Three Solutions, One Platform

All three solutions run on the same Elasticsearch engine. That *one platform, three solutions* story is Elastic's core pitch.

| Solution | Problem it solves | Typical buyer | Examples |
| --- | --- | --- | --- |
| **Search & AI** | Find the right information fast; give AI apps the right context | Developers, product and digital teams | Site and app search, chatbots grounded in company docs, AI agents |
| **Observability** | Understand why apps and infrastructure are slow or broken | SREs, DevOps, IT Ops | Logs, metrics, traces, APM, root-cause analysis |
| **Security** | Detect and respond to cyber threats | SOC teams, CISOs | SIEM, endpoint protection (XDR), cloud security |

**Why one platform matters:** The same data serves operations and security teams, so customers store it once and consolidate several tools.

**Proof points:**
- CISA selected Elastic to power a SIEM-as-a-Service offering for US federal civilian agencies.
- Named a Leader in the Gartner Magic Quadrant for Observability Platforms.

## Deployment Options

| Option | What it is | Best for |
| --- | --- | --- |
| **Elastic Cloud Serverless** | Fully managed; no clusters to size; pay for usage | Fastest start, variable workloads |
| **Elastic Cloud Hosted** | Managed clusters on AWS, Azure, or Google Cloud (also via their marketplaces) | Cloud customers wanting more control |
| **Self-managed** | Customer runs the software in its own data center or cloud | Regulated, on-prem, or air-gapped environments |

**Licensing:** The free core is open source again (AGPLv3 option added November 2024); paid tiers add advanced security, AI features, and support. In 2021 Elastic changed licenses to stop cloud providers from reselling it, and AWS forked the project into **OpenSearch**.

**Government:** Elastic Cloud Hosted achieved FedRAMP High authorization on AWS GovCloud in 2026.

## The AI Angle

Elastic's position: LLMs are only as good as the context they receive, and finding the right context is a search problem. Elastic calls this **context engineering**. As of mid-2026, 37% of its $100K+ customers use its AI features, up from 21% a year earlier.

| Capability | What it does |
| --- | --- |
| **Vector search** | Stores embeddings so users can search by meaning, not just keywords |
| **Hybrid search + reranking** | Combines keyword and semantic search, then reorders for the best results |
| **ELSER and Jina AI models** | Elastic's own models for semantic search and embeddings |
| **Elastic Inference Service (EIS)** | GPU-powered model hosting inside Elastic |
| **Agent Builder** | Build AI agents that work with company data (generally available since January 2026) |
| **MCP and A2A support** | Connects Elastic to external AI agents and tools through open standards |
| **Workflows** | Native automation within the platform (GA in Elastic 9.4) |
| **AI in Observability and Security** | Agentic features that speed up root-cause analysis and threat triage |

**Key partners:** AWS, Microsoft Azure (retrieval layer for Azure AI Foundry agents), Google Cloud (2026 Partner of the Year, Marketplace), NVIDIA, Dell.

## Competitive Landscape

| Market | Main competitors |
| --- | --- |
| Search & AI / vector databases | OpenSearch, Algolia, Pinecone, Weaviate, MongoDB Atlas Search |
| Observability | Datadog, Splunk (Cisco), Dynatrace, New Relic, Grafana Labs |
| Security / SIEM | Splunk (Cisco), Microsoft Sentinel, CrowdStrike, Google SecOps, Palo Alto Networks |

**Differentiators:**
- **One platform** for search, observability, and security
- **Search expertise:** 15+ years of relevance work; better retrieval means better AI
- **Run anywhere:** serverless, all major clouds, on-prem, or air-gapped
- **Open and developer-friendly:** large open-source community and a free on-ramp
- **Cost at scale:** efficient storage of large data volumes
