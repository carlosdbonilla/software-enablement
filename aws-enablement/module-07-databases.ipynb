## MODULE 7 INTRODUCTION

### Introduction to Databases
**Relational vs nonrelational:**

| | **Relational (SQL)** | **Nonrelational (NoSQL)** |
|---|---|---|
| Structure | **Tables** of rows and columns with a fixed **schema**; tables linked by relationships (keys) | Flexible models: **key-value, document, graph, wide-column, in-memory** |
| Query language | **SQL** (joins across tables) | Varies by database type |
| Scaling | Usually scale **up** (bigger server) | Built to scale **out** (more servers) |
| Great for | Transactions, structured data, complex queries (orders, finance, CRM) | Massive scale, flexible or fast-changing data, very low latency |

**Unmanaged vs managed databases (Shared Responsibility again):**

| | **Database on EC2 (unmanaged)** | **Amazon RDS (managed)** |
|---|---|---|
| You manage | OS patching, DB install, patching, backups, high availability, scaling | Your data, schema, queries, and access |
| AWS manages | Hardware and infrastructure | Hardware, OS, DB patching, backups, failover |

**Purpose-built databases:** AWS's philosophy is to use the **right database for each job** instead of forcing one database to do everything.

## DATABASES

### Relational Database Services
**Amazon RDS (Relational Database Service)** = managed relational databases.

- **Engines:** Amazon Aurora, PostgreSQL, MySQL, MariaDB, Oracle, Microsoft SQL Server, Db2
- **AWS handles:** hardware provisioning, setup, patching, **automated backups**
- **Multi-AZ deployment:** a **standby copy in another AZ**; automatic failover if the primary fails (high availability)
- **Read replicas:** read-only copies that take on **read-heavy** traffic (scale reads)
- **Encryption** at rest and in transit

**Amazon Aurora** = AWS's cloud-native relational database, **MySQL- and PostgreSQL-compatible**.

- Higher performance than standard MySQL/PostgreSQL at lower cost than commercial databases
- Data is copied **6 ways across 3 AZs**; continuous backup to S3; up to 15 read replicas
- **Aurora Serverless** scales capacity automatically; **Aurora DSQL** is a serverless, distributed option for multi-Region apps

### NoSQL Database Services
**Amazon DynamoDB** = **serverless key-value and document database**.

- **Single-digit millisecond** performance at **any scale**
- No servers to manage; scales automatically; pay per request (**on-demand**) or set capacity (**provisioned**)
- Data model: **tables** → **items** (rows) → **attributes**; each item is found by its **primary key** (partition key); no fixed schema beyond the key
- **Global tables:** replicate across Regions for global, multi-active apps
- Great for: shopping carts, gaming, user profiles, IoT, high-traffic web apps

**RDS vs DynamoDB:**

| Use **RDS/Aurora** when… | Use **DynamoDB** when… |
|---|---|
| You need **complex queries and joins** across related tables | You need **massive scale** with simple, key-based lookups |
| Data has a fixed, well-defined schema | Data shape is flexible or changes often |
| Example: sales and accounting system | Example: real-time shopping cart for millions of users |

### AWS Databases Demonstration
Typical steps in the console:

- **RDS:** choose engine (e.g., PostgreSQL) → template (production / dev-test / free tier) → credentials → instance size → storage → connectivity (VPC, security group) → turn on Multi-AZ for production → create
- **DynamoDB:** create table → table name + **partition key** → choose capacity mode → add items

### In-Memory Caching Services
**Caching** = keeping frequently read data **in memory** so apps get **microsecond** responses and the main database does less work (cheaper and faster).

| Service | What it does | Use cases |
|---|---|---|
| **Amazon ElastiCache** | Managed in-memory cache compatible with **Valkey, Redis OSS, and Memcached** (serverless option available) | Session stores, query caching, gaming leaderboards |
| **Amazon DynamoDB Accelerator (DAX)** | In-memory cache **built for DynamoDB**; read times drop from milliseconds to **microseconds** with minimal app changes | Read-heavy DynamoDB apps |
| **Amazon MemoryDB** | **Durable** in-memory database (Valkey/Redis-compatible) that can be a **primary** database, not just a cache | Ultra-fast apps that can't lose data |

### Additional Database Services

| Service | Type | Use cases |
|---|---|---|
| **Amazon DocumentDB** | **Document** database, **MongoDB-compatible**; stores JSON-style documents | Content management, user profiles, catalogs |
| **Amazon Neptune** | **Graph** database for highly connected data | Social networks, **fraud detection**, recommendation engines |
| **Amazon Keyspaces** | **Wide-column**, Apache Cassandra-compatible | High-scale apps using Cassandra |
| **Amazon Timestream** | **Time-series** database | IoT sensor data, application metrics |
| **AWS Managed Blockchain** | Create and run blockchain networks (Ethereum, Hyperledger Fabric) | Multi-party transactions without a central authority |
| **Amazon QLDB** | **Ledger** database: immutable, verifiable history of every change | Financial records, audit trails |
| **AWS Backup** | Not a database: **centralized backups** across databases and storage | Backup policies, cross-Region copies, compliance |

> ⚠️ **2026 update:** Amazon QLDB **reached end of support on July 31, 2025** (AWS recommends Aurora PostgreSQL instead). It may still appear in course material or older exam questions.

## MODULE 7 CONCLUSION

### Module Summary

- **Relational** = RDS (managed engines) + **Aurora** (cloud-native, high performance)
- **NoSQL** = **DynamoDB** (serverless key-value, single-digit ms at any scale)
- **Caching** = ElastiCache, DAX, MemoryDB
- **Purpose-built** = DocumentDB (document), Neptune (graph), Keyspaces (Cassandra), Timestream (time series)
- Managed databases shift patching, backups, and failover to AWS

## 🎯 Databricks SA Connection

- **OLTP vs OLAP. Know this cold.**
  - **OLTP** (online transaction processing) = many small, fast reads/writes (place an order, update a balance) → **RDS, Aurora, DynamoDB**
  - **OLAP** (online analytical processing) = big queries over lots of history (revenue by region for 5 years) → **data warehouse / lakehouse**
- **These databases are common DATA SOURCES for a lakehouse.** Typical pattern: Aurora/RDS/DynamoDB → ingest changes (**CDC**, change data capture) → Bronze/Silver/Gold tables in Databricks
- **Lakeflow Connect** provides managed connectors to ingest from databases; customers also use **AWS DMS** to land CDC data in S3 for Databricks to process (DMS is covered in Module 12)
- **Lakebase** = Databricks' managed **Postgres** (OLTP) database, which brings operational data into the same platform as analytics and AI
- **Don't run analytics on the production OLTP database.** Big reporting queries slow down the app. That's a classic reason customers adopt a lakehouse
- **Discovery question:** "Which operational systems feed your analytics today, and how fresh does that data need to be?" (answer drives batch vs streaming/CDC)

## 📝 Exam / Interview Quick Hits

- Managed relational = **RDS** | Cloud-native relational = **Aurora** (MySQL/PostgreSQL compatible, 6 copies across 3 AZs)
- **Multi-AZ** = high availability (standby) | **Read replicas** = scale reads
- Serverless NoSQL key-value = **DynamoDB** (single-digit ms) | DynamoDB cache = **DAX** (microseconds)
- Caching = **ElastiCache** | Durable in-memory = **MemoryDB**
- MongoDB = **DocumentDB** | Graph = **Neptune** | Cassandra = **Keyspaces** | Time series = **Timestream**
- Centralized backups = **AWS Backup**
