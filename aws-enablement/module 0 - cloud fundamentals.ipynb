# Cloud Infrastructure: Key Notes

## What It Is
- **Cloud infrastructure** = the hardware + software resources that make up the cloud
- Providers run global **data centers** (servers, storage, networking gear) and handle OS/software setup
- Customers **lease it pay-as-you-go** instead of buying and maintaining hardware

# Cloud Infrastructure: Core Components

## Physical / Global Infrastructure
- **Data Center**: Physical building with servers, storage, networking, power, and cooling
- **Region**: Geographic area containing multiple isolated data center groups (e.g., us-east-1)
- **Availability Zone (AZ)**: One or more data centers within a region, with separate power and networking, used for high availability
- **Edge Location / Point of Presence (PoP)**: Small sites near users that cache content for low latency (used by CDNs)
- **Local Zone**: Extension of a region placed close to large cities for ultra-low latency
- **Wavelength Zone**: Compute embedded in 5G telecom networks
- **Outposts / Hybrid**: Cloud hardware installed in a customer's own data center

## Compute
- **Virtual Machines (VMs)**: Rentable virtual servers (AWS EC2)
- **Containers**: Lightweight packaged apps (Docker; AWS ECS, EKS for Kubernetes)
- **Serverless Functions**: Code that runs on demand with no server management (AWS Lambda)
- **Auto Scaling**: Automatically adds or removes compute based on demand
- **Load Balancer**: Distributes traffic across servers (AWS ELB)
### Compute Resources (What's Inside a Server/VM)
- **CPU (Processor)**: Does the actual computing work
  - **Physical Core**: A real processing unit on the chip
  - **vCPU (Virtual CPU)**: What the cloud sells; usually = 1 thread (hyperthread), so 1 physical core ≈ 2 vCPUs
  - **Architecture**: x86 (Intel/AMD) vs ARM (AWS Graviton, cheaper and more efficient)
- **Memory (RAM)**: Fast, temporary working space for running apps and data; cleared when the machine stops
  - Measured in GiB (e.g., 16 GiB)
  - More RAM = more data processed in-memory (important for Spark/Databricks)
- **GPU / Accelerators**: Specialized processors for AI/ML and graphics (NVIDIA GPUs, AWS Trainium/Inferentia)
- **Local Storage (Instance Store)**: Disk physically attached to the host; fast but temporary
- **Network Bandwidth**: Throughput the instance gets (e.g., up to 12.5 Gbps)

### Virtualization (How One Server Becomes Many)
- **Bare Metal / Host**: The physical server in the data center
- **Hypervisor**: Software that splits one physical server into multiple isolated VMs (AWS Nitro, VMware, KVM)
- **Virtual Machine (VM) / Instance**: A slice of the host with its own allotted vCPU + RAM
- **LPAR (Logical Partition)**: IBM Power/mainframe equivalent of a VM; hardware split into isolated partitions
  - e.g., 8 cores x 16 GB LPAR ≈ cloud instance with 8 vCPU / 16 GiB RAM
- **Container**: Shares the host OS kernel; lighter than a VM, gets CPU/RAM limits instead of dedicated hardware

### Instance Sizing (How Cloud Packages CPU + RAM)
- **Instance Type**: A preset bundle of vCPU, RAM, and network (e.g., AWS `m6i.2xlarge` = 8 vCPU / 32 GiB)
- **Instance Families**: Optimized for different workloads
  - **General Purpose (M, T)**: Balanced CPU:RAM, roughly 1:4
  - **Compute Optimized (C)**: More CPU per GB, roughly 1:2 (e.g., 8 vCPU / 16 GiB)
  - **Memory Optimized (R, X)**: More RAM per CPU, roughly 1:8+ (good for big data/Spark)
  - **Storage Optimized (I, D)**: Fast local disks
  - **Accelerated Computing (P, G)**: GPUs for AI/ML
- **Naming Pattern (AWS)**: `c6i.2xlarge` = family (c) + generation (6) + processor (i = Intel) + size (2xlarge)
- **Scaling Up (Vertical)**: Bigger instance (more CPU/RAM)
- **Scaling Out (Horizontal)**: More instances (how Spark/Databricks clusters scale)

### Pricing Models
- **On-Demand**: Pay by the second/hour, no commitment
- **Reserved / Savings Plans**: 1–3 year commitment for a big discount
- **Spot Instances**: Spare capacity at up to ~90% off, but can be reclaimed (common for Databricks worker nodes)

## Storage
- **Object Storage**: Files stored as objects in buckets; cheap and highly scalable (AWS S3)
- **Block Storage**: Disk volumes attached to VMs (AWS EBS)
- **File Storage**: Shared network file systems (AWS EFS)
- **Archive Storage**: Very low-cost, slow-retrieval long-term storage (AWS S3 Glacier)
- **Backup**: Managed backup and restore (AWS Backup)

## Networking
- **Virtual Private Cloud (VPC)**: Your own isolated private network in the cloud
- **Subnets**: Segments of a VPC; public (internet-facing) or private
- **Internet Gateway**: Connects a VPC to the internet
- **NAT Gateway**: Lets private resources reach the internet without being exposed
- **Route Tables**: Rules for where network traffic goes
- **Security Groups / NACLs**: Virtual firewalls at the instance and subnet level
- **DNS**: Translates domain names to IP addresses (AWS Route 53)
- **CDN**: Delivers content from edge locations (AWS CloudFront)
- **VPN / Direct Connect**: Secure or dedicated links between on-prem and the cloud
- **API Gateway**: Front door for creating and managing APIs

## Databases
- **Relational (SQL)**: Structured tables (AWS RDS, Aurora)
- **NoSQL**: Key-value or document data, very flexible and fast (AWS DynamoDB)
- **In-Memory Cache**: Extremely fast data access (AWS ElastiCache)
- **Data Warehouse**: Optimized for analytics on large datasets (AWS Redshift)
- **Graph / Time-Series**: Specialized databases for relationships or time-based data

## Data & Analytics
- **Data Lake**: Central store for raw data of any type (usually built on object storage)
- **Lakehouse**: Combines a data lake with warehouse features (Databricks, Delta Lake)
- **ETL / Data Integration**: Extract, transform, and load pipelines (AWS Glue)
- **Streaming**: Real-time data ingestion (AWS Kinesis, Kafka/MSK)
- **Query Engines**: SQL on data in the lake (AWS Athena)
- **BI / Visualization**: Dashboards and reporting (AWS QuickSight)

## AI / Machine Learning
- **ML Platform**: Build, train, and deploy models (AWS SageMaker)
- **Generative AI Services**: Access to foundation models / LLMs (AWS Bedrock)
- **GPUs / Accelerators**: Specialized hardware for training and inference

## Security & Identity
- **IAM (Identity & Access Management)**: Users, roles, and permissions
- **Encryption / Key Management**: Protects data at rest and in transit (AWS KMS)
- **Secrets Management**: Securely stores passwords and API keys
- **Threat Detection / Firewalls**: WAF, DDoS protection (AWS Shield, GuardDuty)
- **Compliance / Auditing**: Logs of who did what (AWS CloudTrail)

## Management & Monitoring
- **Monitoring & Logging**: Metrics, logs, and alerts (AWS CloudWatch)
- **Infrastructure as Code (IaC)**: Define infrastructure in code (Terraform, AWS CloudFormation)
- **Cost Management**: Billing, budgets, and cost tracking
- **Governance**: Multi-account management and policies (AWS Organizations)

## Application Integration
- **Message Queues**: Decouple systems with queued messages (AWS SQS)
- **Pub/Sub Notifications**: Broadcast events to subscribers (AWS SNS)
- **Event Bus**: Route events between services (AWS EventBridge)
- **Workflow Orchestration**: Coordinate multi-step processes (AWS Step Functions)

## Developer Tools
- **CI/CD Pipelines**: Automated build, test, and deploy
- **Code Repositories**: Source control (GitHub, AWS CodeCommit)
- **SDKs / CLI**: Tools to interact with cloud services through code or the command line

## Service Models (How It's Sold)
- **IaaS**: You manage the OS and apps; the provider manages hardware (e.g., EC2)
- **PaaS**: You manage only apps and data (e.g., Elastic Beanstalk)
- **SaaS**: Fully managed software (e.g., Salesforce, Databricks)
- **Shared Responsibility Model**: The provider secures the cloud itself; the customer secures what they put in it
## Delivery Models (who manages what)
- **IaaS** (Infrastructure as a Service): rent raw building blocks (servers, storage, networking, OS). *Most customer control.*
- **PaaS** (Platform as a Service): provider manages the dev environment; you just build and deploy apps. *Middle ground.*
- **SaaS** (Software as a Service): finished app in the browser via subscription; provider handles everything. *Least customer effort.*

> Rule of thumb: **IaaS → PaaS → SaaS** = less to manage, less control.

## Adoption Models (where it runs)
- **Public cloud**: rent shared (multi-tenant) infrastructure from a provider. Cheapest, most elastic. Dedicated hardware available at a premium.
- **Private cloud**: one organization owns and runs it (usually on-prem). More control and isolation, higher cost. Can be outsourced as *managed private cloud*.
- **Hybrid cloud**: both at once. Public for scale and global reach, private for sensitive data.

## Infrastructure vs. Architecture
- **Architecture** = the *blueprint*: how the pieces are designed to work together (microservices, APIs, containers)
- **Infrastructure** = the *physical + software resources* that make that design run (compute, storage, connectivity)

## Clarifications (where the source oversimplifies)
- IaaS does **not** give "complete control of the entire stack." The provider still owns and runs the physical hardware; you control the OS and everything above it.
- Cloud resources are not accessed only through a GUI. Engineers also use **APIs, CLIs, and infrastructure-as-code** (for example, Terraform).

## Why This Matters for Databricks
- Databricks is a **PaaS/SaaS-style data platform** that runs **on top of** public cloud IaaS (AWS, Azure, GCP).
- Customer data lives in **cloud object storage** in the customer's own account. Databricks spins up **compute clusters** (VMs) to process it.
- Expect customers to ask about **hybrid/private** setups (for example, sensitive data on-prem) and **networking/security**, such as private connectivity.
