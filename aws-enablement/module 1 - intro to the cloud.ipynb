## 1. What is Cloud Computing?
**Definition (memorize this):** The *on-demand delivery of IT resources over the internet with pay-as-you-go pricing.*

- **On-demand** → get servers, storage, databases in minutes, not weeks
- **Over the internet** → no physical hardware on your side
- **Pay-as-you-go** → pay only for what you use, like an electricity bill

### Deployment Models
| Model | What it means | Example |
|---|---|---|
| **Cloud (all-in)** | Everything runs in the cloud | A startup built 100% on AWS |
| **On-premises (private cloud)** | Company owns and runs its own data center | A bank running its own servers |
| **Hybrid** | Mix of on-prem + cloud, connected together | Keep sensitive data on-prem, run analytics in the cloud |

> 💡 Hybrid is very common in regulated industries (banking, healthcare, government) because of legacy systems and compliance rules.

## 2. Six Benefits of the AWS Cloud
1. **Trade fixed expense for variable expense** → no big upfront hardware purchase (CapEx → OpEx)
2. **Benefit from massive economies of scale** → AWS buys at huge volume, so prices are lower
3. **Stop guessing capacity** → scale up or down based on real demand
4. **Increase speed and agility** → spin up resources in minutes, experiment faster
5. **Stop spending money running data centers** → focus on your business, not racking servers
6. **Go global in minutes** → deploy to other regions around the world quickly

## 3. AWS Global Infrastructure
| Term | What it is |
|---|---|
| **Region** | A geographic area (e.g., `us-east-1` N. Virginia) containing multiple Availability Zones |
| **Availability Zone (AZ)** | One or more data centers with redundant power, networking, and connectivity. Each Region has **at least 3** AZs, physically separated |
| **Edge Location** | Sites closer to end users that cache content for faster delivery (covered more in Module 4) |

**Why it matters:** Running across multiple AZs = **high availability**. If one data center fails, your app keeps running in another.**Fault Tolerance**

## 4. Shared Responsibility Model 🔐
| AWS = Security **OF** the Cloud | Customer = Security **IN** the Cloud |
|---|---|
| Physical data centers & facilities | Your **data** (and encrypting it) |
| Hardware, servers, networking | **Who has access** (IAM, permissions) |
| Global infrastructure (Regions, AZs) | OS patching, firewall & network config (on services like EC2) |
| Virtualization layer | Your **applications** and code |

**Analogy:** AWS is the landlord (secures the building). You're the tenant (lock your own apartment door).

**Key rule:** The more *managed* the service, the more AWS handles.
- EC2 (raw virtual server) → **you** manage more
- Serverless / fully managed services → **AWS** manages more

## 5. Cloud in Real Life
Global infrastructure + shared responsibility together = a business that is fast, resilient, and secure. A small company can get the same global reach as a Fortune 500 without building data centers.

- **Databricks runs on top of AWS** (also Azure and GCP). Customer data typically lives in their own S3 buckets in their own AWS account.
- **Shared responsibility gets a third layer:** AWS secures the infrastructure, Databricks secures its platform, the customer secures their data, access, and governance (Unity Catalog).
- **Regions matter in regulated deals:** data residency and compliance often dictate which AWS Region the workspace must run in.
- **Benefits = your value story:** "stop guessing capacity" maps to Databricks auto-scaling clusters and serverless compute.
