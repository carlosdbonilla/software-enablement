## MONITORING, COMPLIANCE, AND GOVERNANCE

### Introduction to Monitoring (reading)

- **Monitoring** = collecting and analyzing data about your resources to keep them **healthy, performant, secure, and cost-efficient**
- What you collect:
  - **Metrics:** numbers over time (CPU %, request count, latency)
  - **Logs:** detailed records of events (application errors, access logs)
  - **Events/alerts:** notifications when something happens or crosses a threshold
- Benefits: find and fix problems **before users notice**, plan capacity, control cost, support audits

### Amazon CloudWatch
**Real-time monitoring** of AWS resources, applications, and even on-premises servers.

| Feature | What it does |
|---|---|
| **Metrics** | Collects data points from resources (e.g., EC2 CPU utilization) |
| **Alarms** | Act when a metric **crosses a threshold**: send a notification (via **SNS**), trigger **Auto Scaling**, or stop/reboot an instance. Example: stop idle dev instances when CPU stays under 5% |
| **Dashboards** | Customizable views of many metrics in one place |
| **Logs** | Centralizes logs from systems, apps, and AWS services; search and analyze them |

Benefits: one place to see everything, faster troubleshooting (**reduced mean time to resolution**), and lower cost by spotting waste.

### AWS CloudTrail
**Records API calls and user activity** in your account. Answers: **who** did **what**, **when**, and **from where** (source IP).

- **Events:** every API call, console action, and SDK/CLI request
- **Event history:** free, searchable record of the **last 90 days**
- **Trails:** deliver log files to an **S3 bucket** for long-term storage (audit evidence for regulations like PCI and HIPAA)
- **CloudTrail Insights:** learns normal API activity and **alerts on unusual patterns** (e.g., a sudden spike in resource deletions)
- Use cases: **auditing**, security investigations, troubleshooting, proving compliance

> 💡 **CloudWatch vs CloudTrail:** CloudWatch = **"How is it performing?"** (metrics, logs, alarms) | CloudTrail = **"Who did what?"** (API audit trail)

### Compliance

- AWS is audited against many **compliance programs**: SOC 1/2/3, ISO 27001, PCI DSS, HIPAA, FedRAMP, GDPR, and more
- Customers **inherit** AWS's controls for the infrastructure layer, but are still responsible for configuring **their** workloads compliantly (Shared Responsibility)

**AWS Artifact** = free, self-service portal for compliance documents:

| Part | What it gives you |
|---|---|
| **Artifact Reports** | AWS security and compliance reports from **third-party auditors** (e.g., SOC reports) to share with **your** auditors |
| **Artifact Agreements** | Review, accept, and manage agreements with AWS (e.g., a **HIPAA Business Associate Addendum**) for one account or all accounts in your organization |

### Auditing AWS Resources for Compliance

| Service | What it does |
|---|---|
| **AWS Config** | Keeps an **inventory and configuration history** of your resources and continuously checks them against **rules** (e.g., "all S3 buckets must be encrypted", "no security group allows SSH from anywhere"). Flags noncompliant resources and can auto-remediate |
| **AWS Audit Manager** | Continuously **collects audit evidence** and maps it to frameworks (e.g., PCI DSS, HIPAA) to produce **audit-ready reports** |

> 💡 **CloudTrail vs Config:** CloudTrail = **who made the change** | Config = **what the resource looks like now, how it changed, and whether it complies**

### AWS Organizations
**Centrally manage multiple AWS accounts.**

| Concept | What it is |
|---|---|
| **Organization** | Collection of accounts arranged in a tree |
| **Root** | Top-level container of the tree |
| **Management account** | Creates and governs the organization; pays the consolidated bill |
| **Member accounts** | All other accounts in the organization |
| **Organizational units (OUs)** | Groups of accounts (e.g., Prod, Dev, Finance, Security); can be nested |
| **Service control policies (SCPs)** | Guardrails that set the **maximum permissions** for accounts or OUs (e.g., "no one in Dev can use Regions outside the US") |
| **Consolidated billing** | **One bill** for all accounts; combined usage earns **volume discounts** |

> 💡 SCPs **don't grant** permissions. They only **limit** what IAM policies in those accounts can allow.

**Why multiple accounts?** Separate environments (prod vs dev), limit the "blast radius" of mistakes, clearer billing by team, and different security rules per workload.

### Governance
**Governance at scale** = enforcing company rules across many accounts, services, and licenses.

| Service | What it does |
|---|---|
| **AWS Control Tower** | Sets up and governs a secure, multi-account environment (**landing zone**) based on best practices. Includes **controls** (guardrails), an **Account Factory** for standardized new accounts, and a **dashboard** showing noncompliant resources |
| **AWS Service Catalog** | Catalog of **pre-approved** AWS products/templates teams can deploy themselves, staying within company standards |
| **AWS License Manager** | Tracks and manages **software licenses** (including **Bring Your Own License**) to stay compliant and control licensing costs |

### AWS Health (reading)

- Your source for **events that affect the health of your AWS resources**: service issues, **scheduled maintenance**, and account notifications
- **AWS Health Dashboard:** account-specific view of events and recommended actions
- **AWS Health API:** programmatic access (requires Business-level support or higher) and integration with EventBridge for automation
- Use cases: troubleshoot incidents, prepare for planned lifecycle events (e.g., an engine version retiring)

### AWS Trusted Advisor
**Automated best-practice checks** for your AWS environment, with recommendations.

| Category | Checks for… |
|---|---|
| **Cost optimization** | Idle or underused resources, savings opportunities |
| **Performance** | Ways to improve speed and responsiveness |
| **Security** | Open ports, missing MFA on root, overly permissive access |
| **Fault tolerance** | Missing redundancy (e.g., no Multi-AZ, no backups) |
| **Service limits** | Usage approaching **service quotas** |
| **Operational excellence** | Operational best practices (newer category) |

**Status indicators:** 🟢 no problem detected | 🟠 investigation recommended | 🔴 action recommended

**Access depends on support plan:** every plan gets core checks; **Business-level support and above** get the full set.

**Related: IAM Access Analyzer** = identifies resources shared outside your account and **unused or overly broad permissions** to tighten least privilege.

## 🎯 Databricks SA Connection

- **Audit logs and system tables = CloudTrail for the data platform.** Databricks audit logs and **system tables** (audit, billing/usage, lineage, compute) record who accessed which data and what it cost. Security teams usually want **both** CloudTrail and Databricks audit logs feeding their SIEM
- **Compliance in regulated accounts:** Databricks offers a **compliance security profile** for workloads like **HIPAA and PCI-DSS**, and its own reports via the **Databricks Trust Center** (like AWS Artifact). Comes up in almost every regulated deal
- **Governance at two layers:**
  - **AWS** governs **infrastructure** (accounts, networks, resources) → Organizations, SCPs, Control Tower
  - **Unity Catalog** governs **data and AI assets** (tables, files, models), with access control and **lineage** of how data flows
  - They're complementary, not competing
- **Guardrails in Databricks:** **compute policies** limit which instance types, sizes, and settings users can launch (similar spirit to SCPs and Service Catalog)
- **Tags flow down:** Databricks cluster tags propagate to the EC2 instances, so customers can track Databricks infrastructure spend by team in AWS cost tools
- **Discovery question:** "Which compliance frameworks do you have to meet, and who signs off?"

## 📝 Exam / Interview Quick Hits

- **CloudWatch** = metrics, alarms, dashboards, logs ("how is it performing?")
- **CloudTrail** = API audit log ("who did what, when?"); 90-day event history; Insights = unusual activity
- **Config** = resource configuration history + compliance rules | **Audit Manager** = audit evidence
- **Artifact** = download compliance reports + accept agreements
- **Organizations** = multi-account, OUs, **SCPs** (max permissions), consolidated billing
- **Control Tower** = landing zone + guardrails | **Service Catalog** = approved products | **License Manager** = licenses
- **Trusted Advisor** = cost, performance, security, fault tolerance, service limits (+ operational excellence)
