## MODULE 3 INTRODUCTION

### Introduction to Serverless Computing

- **Serverless** = you can't see or access the underlying servers; AWS provisions, scales, and maintains them
- You focus on your **code**, not on infrastructure
- Scales automatically and you pay only for what you use
- **EC2 vs serverless:** EC2 = you manage instances and the OS | Serverless = AWS manages all of that

**Compute spectrum (more control → less management):**
**EC2** → **Containers (ECS/EKS on EC2)** → **Fargate** → **Lambda**

## AWS SERVERLESS, CONTAINERS, AND SOLUTIONS OVERVIEW

### AWS Lambda

- Serverless compute that **runs your code in response to events (triggers)**
- How it works:
  1. Upload code as a **Lambda function**
  2. Set a **trigger** (e.g., file uploaded to S3, API request, message in SQS)
  3. Code runs only when triggered
  4. Pay only for the compute time used (billed per millisecond)
- Max run time per invocation: **15 minutes** → best for short, event-driven tasks (not long-running jobs)

### Demo: Creating an AWS Lambda Function
Steps in the console:

1. Open Lambda → **Create function** → **Author from scratch** (or start from a blueprint)
2. Name the function and pick a **runtime** (e.g., Python, Node.js)
3. Choose or create an **execution role** = the IAM role that lets the function call other services (e.g., write logs to CloudWatch)
4. Write or paste code in the built-in editor → **Deploy**
5. **Add a trigger** (e.g., S3 upload, API Gateway, EventBridge schedule)
6. **Test** with a sample event → check the output and logs in **CloudWatch Logs**

> 💡 Two settings you can tune: **memory** (CPU scales with it) and **timeout** (up to 15 minutes).

### Containers and Orchestration on AWS

- **Container** = a standard package of your code + all its dependencies, so it runs the same everywhere (laptop, test, production)
- **Docker** is the most common container platform
- **Orchestration** = managing many containers at scale (deploy, start/stop, scale, monitor, restart failed containers)

| Service | What it is |
|---|---|
| **Amazon ECS** (Elastic Container Service) | AWS's own highly scalable container orchestration service |
| **Amazon EKS** (Elastic Kubernetes Service) | Fully managed **Kubernetes** (open-source orchestration standard) |
| **AWS Fargate** | **Serverless** compute engine for containers; works with ECS and EKS, no servers to manage |
| **Amazon ECR** (Elastic Container Registry) | Stores and manages your container images |

**Two ways to run containers:** on **EC2** (you manage the instances, more control) or on **Fargate** (AWS manages it, serverless)

### Additional Compute Services

| Service | What it does | Use it when… |
|---|---|---|
| **AWS Elastic Beanstalk** | Upload your code; AWS handles capacity, load balancing, scaling, and health monitoring (PaaS) | You want to deploy web apps without managing infrastructure |
| **AWS Batch** | Plans, schedules, and runs batch computing jobs at scale | Running large volumes of batch jobs |
| **Amazon Lightsail** | Easy virtual servers, storage, and networking at a low, predictable monthly price | Simple websites, small apps, small businesses |
| **AWS Outposts** | AWS infrastructure and services running in **your own data center** | Hybrid workloads needing low latency or local data processing |

## MODULE 3 CONCLUSION

### Module Summary

- **Serverless** = no servers to manage, automatic scaling, pay per use
- **Lambda** = event-driven functions, up to 15 minutes
- **Containers** package apps consistently; **ECS/EKS** orchestrate them; **Fargate** runs them serverless; **ECR** stores images
- **Beanstalk, Batch, Lightsail, Outposts** cover specialized compute needs

## 🎯 Databricks SA Connection

- **Databricks Serverless compute** follows the same model as Lambda/Fargate: no clusters to manage, fast startup, pay for what runs
- **Lambda as pipeline "glue":** customers often use Lambda triggers on new S3 files. In Databricks, **Auto Loader** and **file arrival triggers** in Lakeflow Jobs can replace much of that custom code
- **EKS/containers:** customers often serve ML models or apps on EKS. **Databricks Model Serving** and **Databricks Apps** are managed alternatives inside the platform
- **AWS Batch ↔ Lakeflow Jobs:** both schedule and run batch work; Jobs adds data-aware orchestration and governance via Unity Catalog
- **Discovery question:** "How much infrastructure does your team want to own and operate?" The answer tells you how strongly to lead with serverless

## 📝 Exam / Interview Quick Hits

- Serverless = no server management, pay per use → **Lambda**, **Fargate**
- Lambda: event-triggered, max **15 min**, billed per ms; needs an **execution role** to call other services
- **ECS** = AWS orchestration | **EKS** = Kubernetes | **Fargate** = serverless containers | **ECR** = image registry
- **Elastic Beanstalk** = upload code, AWS runs it
- **Batch** = batch jobs | **Lightsail** = simple, low-cost servers | **Outposts** = AWS in your data center
