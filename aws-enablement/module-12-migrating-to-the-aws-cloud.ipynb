## MODULE 12 INTRODUCTION

### Introduction to Migration

- **Migration** = moving on-premises (or other-cloud) applications, data, and infrastructure to AWS
- Why companies migrate: **lower cost**, **agility**, **scalability**, **innovation** (AI, analytics), exiting a data center or ending a hardware refresh
- Migrations are usually **planned and phased**, not done in one big bang

**AWS's three-phase migration process:**

| Phase | Goal |
|---|---|
| **1. Assess** | Understand your starting point and build the **business case** (cost, readiness) |
| **2. Mobilize** | Close gaps: detailed **plan**, inventory of apps and dependencies, skills, landing zone |
| **3. Migrate and Modernize** | Move workloads in waves, then **optimize/modernize** them |

## MIGRATION

### AWS Cloud Adoption Framework
**AWS CAF** = guidance that helps organizations plan a successful move to the cloud. It's organized into **six perspectives**, each with its own stakeholders:

| Perspective | Focus | Typical stakeholders |
|---|---|---|
| **Business** | Cloud investments drive **business outcomes** (revenue, cost, agility) | CEO, CFO, COO, business leaders |
| **People** | Culture, org structure, **skills and training**, change management | HR, leaders, people managers |
| **Governance** | Manage **risk**, compliance, budgets, portfolio and program management | CIO, CFO, CDO, risk officers |
| **Platform** | **Architecture** of a scalable cloud platform (hybrid, landing zone) | CTO, architects, engineers |
| **Security** | **Confidentiality, integrity, and availability** of data and workloads | CISO, security architects |
| **Operations** | **Run and support** cloud workloads at agreed service levels | IT operations, SREs, service managers |

**Memory trick:** first 3 = **business** capabilities (**B**usiness, **P**eople, **G**overnance) | last 3 = **technical** capabilities (**P**latform, **S**ecurity, **O**perations)

**CAF transformation journey:** **Envision** (identify opportunities) → **Align** (gaps and dependencies across teams) → **Launch** (pilot projects that deliver value) → **Scale** (expand to production at scale)

### Seven Migration Strategies
The **7 Rs** (simplest → most transformative):

| Strategy | What it means | Example |
|---|---|---|
| **Retire** | Turn it off; no longer needed | An old reporting app nobody uses |
| **Retain** | Keep on-premises **for now** (revisit later) | App that depends on hardware or a contract that hasn't ended |
| **Rehost** ("lift and shift") | Move **as-is**, no changes | On-prem VM → EC2 |
| **Relocate** | Move to AWS at the **hypervisor level**, no new hardware or app changes | VMware environment → VMware Cloud on AWS |
| **Repurchase** ("drop and shop") | Replace with a **different product**, often SaaS | On-prem CRM → Salesforce |
| **Replatform** ("lift, tinker, and shift") | A few **cloud optimizations** without changing core architecture | Self-managed MySQL → Amazon RDS |
| **Refactor / Re-architect** | **Rebuild** using cloud-native features | Monolith → serverless microservices |

> 💡 Rehost is fastest; **Refactor** takes the most effort but delivers the most cloud benefit. Companies often rehost first, then modernize later.

### Migration Services and Tools

| Tool | Phase | What it does |
|---|---|---|
| **Migration Evaluator** | Assess | Builds a data-driven **business case**: current footprint, projected AWS costs |
| **AWS Application Discovery Service** | Mobilize | **Inventories** on-prem servers, their **dependencies**, configuration, and performance |
| **AWS Migration Hub** | Mobilize → Migrate | Central place to **track** discovery, planning, and migration progress |
| **AWS Application Migration Service** | Migrate | Automates **rehosting** servers to AWS with continuous replication and quick cutover |
| **AWS Transform** | Migrate and Modernize | **AI-powered** service (launched 2025) that automates migration and modernization of legacy apps and code |

> ⚠️ **2026 update:** **Migration Hub** and **Application Discovery Service** closed to **new** customers on **Nov 7, 2025**. AWS now points customers to **AWS Transform**. The course may still teach them, so know both.

### Database Migrations

- A database migration is a chance to **modernize** (e.g., move from expensive commercial licenses to managed or open-source engines)
- **Homogeneous** = same engine (MySQL → Amazon RDS for MySQL). Straightforward
- **Heterogeneous** = different engines (Oracle → Aurora PostgreSQL). Requires **converting the schema and code** first

| Service | What it does |
|---|---|
| **AWS Database Migration Service (DMS)** | Migrates databases and data warehouses to AWS **while the source stays running** (minimal downtime). Supports **ongoing replication** of changes (CDC) |
| **AWS Schema Conversion Tool (SCT)** / **DMS Schema Conversion** | Converts **schemas and code objects** (stored procedures, views, functions) from one engine to another for heterogeneous migrations; estimates the effort |

**Heterogeneous migration in 2 steps:** (1) **convert** the schema with SCT/DMS Schema Conversion → (2) **move the data** with DMS

**DMS use cases:** move to managed databases, cut licensing costs, consolidate databases, continuous replication (e.g., feed a data lake), dev/test copies.

### Transferring Data Online (reading)

| Service | What it does |
|---|---|
| **AWS DataSync** | Automates and accelerates moving **large amounts of file/object data** between on-prem storage (NFS, SMB, HDFS, object) and AWS (S3, EFS, FSx) |
| **AWS Transfer Family** | Managed **SFTP, FTPS, and FTP** file transfers into and out of S3 and EFS (keep existing partner file-transfer workflows) |
| **AWS Direct Connect** | Dedicated private network connection for fast, consistent transfers (Module 5) |

### Transferring Data Offline (reading)
When the network is too slow for the amount of data (e.g., petabytes over a limited connection), ship it on **physical devices**:

| Option | What it is |
|---|---|
| **AWS Snowball Edge** | Rugged device you load with data and ship back to AWS. **Storage Optimized** (210 TB) for data transfer; **Compute Optimized** for edge computing in remote locations |
| **AWS Snowcone** | Small, portable Snow device (**discontinued** Nov 2024) |
| **AWS Snowmobile** | Truck-sized data transfer for exabytes (**retired**) |
| **AWS Data Transfer Terminal** | Physical locations where you bring your own storage and upload to AWS at high speed |

> ⚠️ **2026 update:** since **Nov 7, 2025**, Snowball Edge is available **only to existing customers**. AWS now recommends **DataSync** for most migrations and **Data Transfer Terminal** for secure physical transfers. Expect "Snow Family" in exam questions anyway.

## 🎯 Databricks SA Connection

- **Migrations are a huge part of the Databricks SA job.** Common sources: on-prem **Hadoop/Cloudera**, legacy warehouses (**Teradata, Oracle, Netezza, SQL Server**), **Amazon EMR**, **Redshift**, and Snowflake
- **Map the 7 Rs to data platforms:**
  - Hadoop/EMR Spark jobs → Databricks = mostly **Replatform** (Spark code moves with tweaks) or **Refactor** (rebuild as Lakeflow pipelines)
  - Legacy ETL tools (e.g., Informatica) → Databricks = **Repurchase / Refactor**
  - Legacy warehouse → Databricks SQL = **Replatform** (SQL conversion) plus data movement
- **Lakebridge** = Databricks' migration toolkit: **assess** the legacy estate, **convert** SQL/ETL code, and **validate** the migrated data. Same idea as Migration Evaluator + SCT + DMS, but for analytics workloads
- **Database sources:** **DMS** or **Lakeflow Connect** can stream CDC from operational databases into the lakehouse
- **Bulk data moves:** **DataSync** is a common way to move HDFS/NAS data into S3 before Databricks takes over
- **CAF is great discovery language.** A migration is never just technical. Ask about **People** (skills, training), **Governance** (compliance, risk), and **Business** (what outcome justifies the move?). This maps directly to **MEDDPICC**: Metrics, Decision criteria, Identify pain, Champion
- **Phased approach:** recommend a **pilot use case** first (CAF's "Launch"), prove value, then scale, rather than migrating everything at once
