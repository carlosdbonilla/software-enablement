### AWS Well-Architected Framework
Best practices for designing and running workloads in the cloud, organized into **6 pillars**:

| Pillar | Key question | Example practices |
|---|---|---|
| **Operational Excellence** | Can we run, monitor, and **continuously improve** it? | Infrastructure as code, small reversible changes, learn from failures |
| **Security** | Are data and systems **protected**? | Least privilege, encryption at rest and in transit, security at every layer, traceability |
| **Reliability** | Does it **recover** from failure and meet demand? | Multi-AZ deployments, automatic recovery, test recovery procedures, scale horizontally |
| **Performance Efficiency** | Are we using the **right resources** efficiently? | Right instance types, serverless, go global in minutes, experiment often |
| **Cost Optimization** | Are we avoiding **unnecessary cost**? | Pay only for what you use, rightsize, measure spend, remove idle resources |
| **Sustainability** | Are we minimizing **environmental impact**? | Maximize utilization, use managed services, remove unused resources |

**Memory trick:** "**OS-PRCS**" → **O**perational excellence, **S**ecurity, **P**erformance efficiency, **R**eliability, **C**ost optimization, **S**ustainability

**Worked example (online flower shop):**

- **Security:** IAM least privilege; encrypt orders and customer data
- **Reliability:** EC2 Auto Scaling across multiple AZs; RDS Multi-AZ
- **Performance:** CloudFront serves product images from S3 quickly worldwide; Compute Optimizer checks instance sizes
- **Cost:** Savings Plans for steady traffic, Spot for flexible jobs, Budgets and Cost Explorer to track spend
- **Operational excellence:** CloudWatch monitoring and alarms; deploy with CloudFormation
- **Sustainability:** scale down off-peak, delete unused resources

**AWS Well-Architected Tool** (free, in the console): answer questions about a workload → get a report of **high- and medium-risk issues** with an **improvement plan**. Supports **milestones** (track progress) and **lenses** (extra guidance for specific workload types, e.g., serverless, ML, data analytics).

**Well-Architected Review** = a structured conversation (often with an AWS Solutions Architect or partner) that assesses a workload against the pillars.

### Specialized Use Cases
Three example architectures that **combine services** to solve a business problem:

**1. Serverless web backend monitored by X-Ray**
`User → Amazon API Gateway → AWS Lambda → Amazon DynamoDB`

- API Gateway receives and validates requests and triggers Lambda; Lambda reads/writes DynamoDB
- **X-Ray** traces each request end to end to find slow or failing steps
- No servers to manage; scales automatically

**2. Serverless static website with a contact form**
`User → static site on Amazon S3 → form submit → API Gateway → Lambda → Amazon SES → email to business owner`

- S3 hosts the website files (no web server); Lambda processes the form; SES sends the email

**3. Customer support with a callback option**

- Customers call in; **Amazon Connect** handles the call with an automated phone menu (**IVR**, interactive voice response) and routes to agents
- During long waits, customers can request a **callback** or switch to **text messaging** instead of holding; **Lambda** runs the logic behind these options

> 💡 The pattern: no single service solves a business problem. **Architecture = choosing and connecting the right services**, which is exactly the SA skill.

## 🎯 Databricks SA Connection

- **Databricks has its own "Well-Architected Lakehouse" framework** (in the Databricks docs). It reuses AWS's pillars and adds lakehouse-specific ones: **data and AI governance** and **interoperability and usability**. Read it right after this module
- **Use the pillars to structure architecture interviews.** After drawing your ingest → consumption design, walk through it pillar by pillar:
  - **Security:** Unity Catalog permissions, PrivateLink, customer-managed keys, audit logs
  - **Reliability:** multi-AZ compute, job retries, Delta Lake ACID transactions and time travel, DR plan across Regions
  - **Performance efficiency:** right compute types, Photon, serverless, liquid clustering
  - **Cost optimization:** serverless/autoscaling, Spot workers, auto-termination, compute policies, billing system tables
  - **Operational excellence:** Asset Bundles or Terraform, CI/CD, monitoring and alerts
  - **Sustainability:** serverless and autoscaling avoid idle compute
- **CI/CD for data:** CodePipeline/CodeBuild (or GitHub Actions) + **Databricks Asset Bundles** is a common pattern for promoting pipelines from dev → test → prod
- **IoT data is a classic streaming use case:** devices → IoT Core → Kinesis → Databricks Structured Streaming → real-time dashboards and ML (e.g., predictive maintenance)
- **Specialized use cases ↔ Databricks Apps:** small data-driven apps can run inside Databricks with governance built in
- **This is the "senior" signal:** hiring managers want to hear **trade-offs** ("we chose X over Y because…"), not just a list of services
