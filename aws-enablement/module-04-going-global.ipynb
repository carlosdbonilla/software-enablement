## MODULE 4 INTRODUCTION

### Introduction to Going Global

- Module 1 introduced **Regions** and **Availability Zones**. This module goes deeper: how to **pick** a Region, what else makes up the global infrastructure, and how to **automate** deployments
- Core goal: **high availability** (your app keeps running when something fails) and **low latency** (fast for users wherever they are)
- Key idea: if a single data center or AZ fails, a well-designed app keeps running in another one

## GOING GLOBAL

### Choosing AWS Regions
Four factors to weigh when selecting a Region:

| # | Factor | What it means | Example |
|---|---|---|---|
| 1 | **Compliance** | Data governance and legal requirements may dictate where data can live | A regulation requires customer data to stay in a specific country |
| 2 | **Proximity** | Closer to customers = lower **latency** | A company with most users in Japan picks the Tokyo Region |
| 3 | **Feature availability** | Not every service is available in every Region (new services roll out gradually) | A new AI service may launch in only a few Regions first |
| 4 | **Pricing** | The same service can cost different amounts in different Regions (taxes, operating costs) | Running the same instance costs more in some Regions |

> 💡 **Compliance usually comes first.** If a law says the data must stay in-country, the other three factors don't matter.

### Diving Deeper into AWS Global Infrastructure

| Component | What it is | Why it matters |
|---|---|---|
| **Region** | A geographic area containing multiple, isolated AZs | Regions are isolated from each other; data stays in a Region unless you choose to move it |
| **Availability Zone (AZ)** | One or more discrete data centers with redundant power, networking, and connectivity. Each Region has **at least 3** | AZs are physically separated (to survive local disasters) but connected with low-latency links |
| **Edge location** | A site **separate from Regions**, close to end users, that caches content | Powers **Amazon CloudFront** (CDN) and **Amazon Route 53** (DNS) for faster delivery |
| **AWS Outposts** | AWS-managed racks installed in **your own data center** | Hybrid workloads that need low latency or local data processing |
| **Local Zones** | AWS compute and storage placed near large cities | Ultra-low latency for users in that metro area |
| **Wavelength Zones** | AWS compute embedded in telecom 5G networks | Ultra-low latency for mobile/5G apps |

**Amazon CloudFront (CDN) in 30 seconds:**

1. A user requests content (image, video, web page)
2. The **nearest edge location** checks its cache
3. **Cache hit** → served immediately | **Cache miss** → fetched from the **origin** (e.g., S3 or EC2), then cached for the next user

**Best practices:**

- Run workloads across **at least 2 AZs** for high availability
- Use **multiple Regions** for disaster recovery or global users

### Infrastructure and Automation
Recap from Module 2: every action in AWS is an **API call**, made through the **Console, CLI, or SDKs**. Clicking through the console doesn't scale and leads to mistakes, so we **automate**.

**Infrastructure as Code (IaC)** = define your infrastructure in text files (templates) instead of clicking.

| Service | What it does |
|---|---|
| **AWS CloudFormation** | Define resources in **JSON or YAML templates**. CloudFormation builds them as a **stack**, in the right order, and can roll back if something fails |
| **AWS Elastic Beanstalk** | (from Module 3) Upload code; AWS provisions EC2, load balancing, and scaling for you |

**Why IaC matters:**

- **Repeatable:** deploy the identical environment in another Region or account in minutes
- **Version-controlled:** templates live in Git, so changes are reviewed and tracked
- **Fewer human errors:** no forgotten checkbox in the console
- **Declarative:** you describe *what* you want; CloudFormation figures out *how*

> 💡 Related tools you'll hear about: **AWS CDK** (write infrastructure in Python/TypeScript, which generates CloudFormation) and **Terraform** by HashiCorp (the most common multi-cloud IaC tool).

## MODULE 4 CONCLUSION

### Module Summary

- Pick a Region based on **compliance, proximity, feature availability, and pricing**
- Global infrastructure = **Regions → AZs → data centers**, plus **edge locations** (CloudFront, Route 53), **Outposts**, **Local Zones**, and **Wavelength**
- Deploy across multiple AZs for high availability
- Automate with **Infrastructure as Code** → **AWS CloudFormation**

## 🎯 Databricks SA Connection

- **Region choice = workspace placement.** A Databricks workspace is deployed in one AWS Region, normally the **same Region as the customer's S3 data** (avoids cross-Region data transfer fees and latency)
- **Regulated customers care about data residency first.** Region and compliance come up early in discovery, just like factor #1 above
- **Feature availability applies to Databricks too:** not every Databricks feature (e.g., certain serverless or AI features) is in every Region. Check before promising
- **Multi-Region = disaster recovery conversations.** Customers ask how to fail over a workspace to another Region (replicating data, code, and configuration)
- **IaC is how platform teams deploy Databricks:** the official **Databricks Terraform provider** creates workspaces, clusters, and permissions; **Databricks Asset Bundles** apply the same "define it as code" idea to jobs and pipelines
- **Discovery question:** "Where does your data live today, and do you have any data residency requirements?"

## 📝 Exam / Interview Quick Hits

- Region selection: **compliance, proximity, feature availability, pricing**
- Each Region has **at least 3 AZs**; deploy across **2+ AZs** for high availability
- **Edge locations** are separate from Regions and power **CloudFront** (CDN) and **Route 53** (DNS)
- **Outposts** = AWS in your data center | **Local Zones** = AWS near big cities | **Wavelength** = AWS in 5G networks
- **CloudFormation** = AWS-native Infrastructure as Code (JSON/YAML templates → stacks)
