## MODULE 9 INTRODUCTION

### Introduction to Security on AWS

- **Shared Responsibility Model (recap from Module 1):**
  - **AWS = security OF the cloud:** physical data centers, hardware, network, virtualization
  - **Customer = security IN the cloud:** data, identities and access, OS and app configuration, encryption, network settings
- How much you own **depends on the service**: more with EC2 (you patch the OS), less with managed/serverless services (S3, Lambda, DynamoDB)
- **Defense in depth** = layers of security (identity, network, application, data, monitoring) so one failure doesn't expose everything
- This module follows the security lifecycle: **prevent** unauthorized access → **protect** networks, apps, and data → **detect and respond** to incidents

## SECURITY

### Preventing Unauthorized Access

- **Authentication** = proving **who you are** (password, MFA) | **Authorization** = what you're **allowed to do** (permissions)
- **AWS IAM (Identity and Access Management)** = manages identities and access to AWS. **Everything is denied by default** until explicitly allowed

| Concept | What it is | Key point |
|---|---|---|
| **Root user** | Identity created with the account; **full access to everything** | Lock it down: strong password + **MFA**; **don't use it for daily tasks** |
| **IAM user** | Identity for **one person or application**, with long-term credentials | Starts with **no permissions**; one user per person |
| **IAM group** | Collection of users (e.g., "Developers", "Finance") | Attach a policy to the group → every member gets it |
| **IAM policy** | **JSON document** that **allows or denies** actions on resources | Attach to users, groups, or roles |
| **IAM role** | Identity you **assume** to get **temporary** permissions; no long-term password | Used by AWS services (e.g., EC2 reading S3), other accounts, and federated/SSO users |
| **MFA** | Multi-factor authentication: password + a second factor (e.g., authenticator app) | Enable for root and all human users |

**Example IAM policy** (allow reading one S3 bucket):
```json
{
  "Version": "2012-10-17",
  "Statement": [{
    "Effect": "Allow",
    "Action": ["s3:GetObject", "s3:ListBucket"],
    "Resource": ["arn:aws:s3:::sales-data", "arn:aws:s3:::sales-data/*"]
  }]
}
```
**Effect** (Allow/Deny) + **Action** (what) + **Resource** (on which thing). An explicit **Deny** always wins.

**Principle of least privilege** = grant **only** the permissions needed to do the job, nothing more.

**Related services:**

| Service | What it does |
|---|---|
| **AWS IAM Identity Center** | **Single sign-on (SSO)** across multiple AWS accounts; connects to corporate identity providers (e.g., Okta, Microsoft Entra ID) |
| **AWS Secrets Manager** | Stores, manages, and **automatically rotates** credentials, API keys, and database passwords (no hard-coded secrets) |
| **AWS Systems Manager** | Central view and management of servers; automates tasks like **patching** |

### IAM Demonstration
Typical steps in the console:

1. Sign in as an admin (not root) → open **IAM**
2. **Create a group** (e.g., "Analysts") and **attach a policy** (e.g., read-only S3)
3. **Create a user** and **add it to the group** (user inherits the group's permissions)
4. **Enable MFA** for the user
5. **Create a role** for a service (e.g., let EC2 read from S3) and attach it to the instance
6. Test: the user can do what the policy allows and nothing else

### Protecting Networks and Applications

- **DoS (denial of service)** = flooding an app with traffic so real users can't reach it
- **DDoS (distributed DoS)** = the same attack launched from **many** compromised machines (a botnet) at once, which makes it harder to block

**AWS layers of protection:**

| Protection | What it does |
|---|---|
| **AWS global infrastructure** | Massive Region and edge capacity absorbs many attacks |
| **Security groups** | Only allow the traffic you specify (e.g., HTTPS only) |
| **Elastic Load Balancing** | Absorbs and spreads incoming traffic before it reaches servers |
| **AWS Shield Standard** | **Free, automatic** protection for **all** AWS customers against common DDoS attacks |
| **AWS Shield Advanced** | **Paid**: detailed attack diagnostics, stronger protection for sophisticated attacks, 24/7 response team, DDoS cost protection; integrates with **CloudFront, Route 53, ELB** |
| **AWS WAF** (Web Application Firewall) | Inspects **web requests** and blocks malicious ones (bad IPs, SQL injection, cross-site scripting, request floods) using rules in a **web ACL** |

> 💡 **Shield** = protects against **DDoS volume** attacks | **WAF** = filters **malicious web requests**. They're often used together.

### Protecting Data

- **Encryption** = scrambling data so it can't be read without the right **key**
  - **At rest** = data stored on disk (S3, EBS, RDS, DynamoDB)
  - **In transit** = data moving across a network, protected with **SSL/TLS** (HTTPS)

| Service | What it does |
|---|---|
| **AWS KMS** (Key Management Service) | Create, manage, and control **encryption keys**; integrates with most AWS services; key usage is logged in CloudTrail |
| **AWS CloudHSM** | Dedicated hardware security modules when you must **fully control** your keys in single-tenant hardware |
| **AWS Certificate Manager (ACM)** | Provision and manage **SSL/TLS certificates** for encryption in transit |
| **Amazon Macie** | Uses ML to **discover and protect sensitive data (PII)** in S3 |
| **AWS Secrets Manager** | Protects passwords, API keys, and credentials (from the previous lesson) |

### Detecting and Responding to Security Incidents

| Service | Role | What it does |
|---|---|---|
| **Amazon Inspector** | **Find vulnerabilities** | Automatically scans **EC2 instances, container images, and Lambda functions** for software vulnerabilities and exposure, with fix recommendations |
| **Amazon GuardDuty** | **Detect threats** | Intelligent threat detection using ML on account activity, network traffic, and DNS logs (e.g., known malicious IPs, crypto-mining) |
| **Amazon Detective** | **Investigate** | Analyzes and visualizes security data to find the **root cause** of a finding |
| **AWS Security Hub** | **Centralize** | One dashboard aggregating findings from many services, checks against security standards, can automate fixes |

**How they fit together:** Inspector finds **weaknesses** → GuardDuty spots **active threats** → Security Hub **collects** everything in one place → Detective helps you **investigate** what happened.

## 🎯 Databricks SA Connection

- **Security is usually the gating item in regulated deals** (banking, healthcare, government). Expect an InfoSec review with detailed questionnaires
- **IAM roles are how Databricks reaches customer data and infrastructure:**
  - A **cross-account IAM role** lets Databricks launch classic compute in the customer's AWS account
  - **Unity Catalog storage credentials** use IAM roles to read/write S3 (no access keys stored anywhere)
- **Identity:** Databricks integrates with corporate **SSO and SCIM** user provisioning (Okta, Microsoft Entra ID), the same idea as IAM Identity Center for AWS
- **Customer-managed keys (KMS):** regulated customers often require **their own keys** for data, workspace storage, and notebooks. Databricks supports this
- **PII governance:** Macie finds PII in S3 files. **Unity Catalog** controls who sees it in tables with **row filters, column masks, tags, and attribute-based access control (ABAC)**
- **Least privilege applies in Databricks too:** grant access on catalogs/schemas/tables only as needed; use groups, not individual users
- **Secrets:** Databricks secret scopes keep credentials out of notebooks (same purpose as Secrets Manager)
- **Discovery question:** "What does your security team need to approve a new data platform?" Get ahead of the InfoSec review early
