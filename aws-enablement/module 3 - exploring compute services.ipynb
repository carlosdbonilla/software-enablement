## 1. The Compute Spectrum
More control ⟵ ⟶ Less management

**EC2 (VMs)** → **Containers (ECS/EKS)** → **Fargate (serverless containers)** → **Lambda (serverless functions)**

> 💡 The further right you go, the more AWS manages and the less you control (Shared Responsibility again).

---

## 2. Serverless: AWS Lambda
- **Serverless** = you don't provision or manage servers; AWS handles capacity and scaling
- **Lambda** runs your code in response to an **event** (file uploaded to S3, API call, message in a queue)
- Pay only for the compute time your code actually runs (per millisecond)
- Max run time per invocation: **15 minutes** → good for short, event-driven tasks

---

## 3. Containers
**Container** = a package of your code + everything it needs to run, so it behaves the same everywhere (laptop, test, prod).

| Service | What it is |
|---|---|
| **Amazon ECS** (Elastic Container Service) | AWS's own container orchestration service |
| **Amazon EKS** (Elastic Kubernetes Service) | Managed **Kubernetes** (open-source orchestration standard) |
| **AWS Fargate** | Serverless engine for containers; works with ECS and EKS, no servers to manage |
| **Amazon ECR** (Elastic Container Registry) | Stores your container images |

**Orchestration** = managing many containers at scale (starting, stopping, scaling, healing).

---

## 4. Other Compute Options
| Service | Use it when… |
|---|---|
| **AWS Elastic Beanstalk** | You upload code; AWS provisions EC2, load balancing, scaling for you (PaaS) |
| **Amazon Lightsail** | Simple, low-cost virtual servers for small apps/websites (fixed monthly price) |
| **AWS Batch** | Running large numbers of batch computing jobs |
| **AWS Outposts** | Running AWS infrastructure inside your own data center (hybrid) |

---

## 🎯 Databricks SA Connection
- **Databricks Serverless** follows the same idea as Lambda/Fargate: customer doesn't manage clusters, pays for what runs, near-instant startup.
- **Lambda in data pipelines:** often used as "glue" (e.g., trigger on a new S3 file). In a Databricks architecture, **Auto Loader** or **file arrival triggers** in Jobs can replace a lot of that custom Lambda code.
- **Containers/EKS:** customers often run apps or ML model serving on EKS. Databricks **Model Serving** and **Databricks Apps** are the managed alternative inside the platform.
- **Positioning tip:** "How much infrastructure does your team want to own?" is a great discovery question; the answer tells you whether serverless is a strong fit.

## 📝 Exam / Interview Quick Hits
- Serverless = no servers to manage, pay per use → **Lambda**, **Fargate**
- Lambda: event-driven, max **15 min**, billed per ms
- ECS = AWS containers | EKS = Kubernetes | Fargate = serverless for both | ECR = image storage
- Elastic Beanstalk = upload code, AWS runs it
- Outposts = AWS in your data center
