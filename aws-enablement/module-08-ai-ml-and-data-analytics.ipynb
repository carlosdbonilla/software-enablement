## MODULE 8 INTRODUCTION

### Introduction to AI and Machine Learning

- **Artificial intelligence (AI)** = machines performing tasks that normally need human intelligence (understanding language, recognizing images, making decisions)
- **Machine learning (ML)** = a subset of AI where systems **learn patterns from data** instead of being explicitly programmed with rules
- **Deep learning** = a subset of ML using multi-layered **neural networks**; powers image recognition, speech, and modern GenAI
- **Generative AI** = models that **create new content** (text, images, code, audio)

**AI ⊃ ML ⊃ Deep learning ⊃ Generative AI**

**Three ways machines learn:**

| Type | How it learns | Example |
|---|---|---|
| **Supervised** | From **labeled** examples (input + correct answer) | Predict whether a transaction is fraud from past labeled transactions |
| **Unsupervised** | Finds **patterns/groups** in unlabeled data | Segment customers into groups by behavior |
| **Reinforcement** | Trial and error with **rewards** | Robot or game agent learning the best moves |

**Key terms:**

- **Training** = teaching the model using historical data | **Inference** = using the trained model to make predictions on new data
- **ML lifecycle:** define the business problem → collect & prepare data → train → evaluate → deploy → **monitor** (and retrain as data changes)

## AI/ML AND DATA ANALYTICS

### AI/ML on AWS
AWS organizes its AI/ML offerings into **three layers**, from easiest to most control:

| Layer | Who it's for | What you get |
|---|---|---|
| **AI services** (pre-trained) | Developers with **no ML expertise** | Ready-to-use APIs: call them and get results |
| **ML services** | Data scientists building **custom models** | **Amazon SageMaker AI**: build, train, and deploy models with managed infrastructure |
| **ML frameworks & infrastructure** | ML experts who want **full control** | Frameworks (PyTorch, TensorFlow) on ML-optimized compute (GPU instances, AWS Trainium/Inferentia chips) |

### AWS AI/ML Solutions (reading)
**Pre-trained AI services (no ML knowledge needed):**

| Service | What it does | Example |
|---|---|---|
| **Amazon Comprehend** | NLP: sentiment, key phrases, entities, language | Analyze customer reviews |
| **Amazon Polly** | **Text → speech** | Voice for an app |
| **Amazon Transcribe** | **Speech → text** (identifies speakers) | Call center transcripts |
| **Amazon Translate** | Real-time language translation | Multilingual website |
| **Amazon Kendra** | Intelligent **enterprise search** using natural language | Search internal documents |
| **Amazon Rekognition** | **Image and video** analysis (objects, faces, text, scenes) | Content moderation |
| **Amazon Textract** | Extract text, forms, and tables from **documents** | Process invoices and forms |
| **Amazon Lex** | Build **chatbots** (voice and text) | Customer service bot |
| **Amazon Personalize** | **Recommendations** from historical data | "Customers also bought…" |

**Amazon SageMaker AI** = full ML platform: notebooks/IDE for data scientists, no-code option for analysts, managed training and hosting, and automated **MLOps** workflows.

### Introduction to Generative AI on AWS

- **Foundation models (FMs)** = very large models **pre-trained on huge amounts of broad data** that can be adapted to many tasks
- **Large language models (LLMs)** = foundation models for **text** (chat, summarize, write, code)
- **Prompt** = the instruction or question you give the model | **Prompt engineering** = writing prompts that get better results
- **Hallucination** = when a model produces confident but **incorrect** output
- **Ways to customize a model** (simplest → most effort):
  1. **Prompt engineering:** better instructions and examples
  2. **RAG (retrieval-augmented generation):** look up relevant company data and add it to the prompt so answers are grounded in facts
  3. **Fine-tuning:** further train the model on your own labeled data
- **Use cases:** chatbots and assistants, document summarization, content creation, code generation, search

### AWS Generative AI Solutions (reading)

| Service | What it does |
|---|---|
| **Amazon Bedrock** | Fully managed access to **foundation models from Amazon and other AI companies** (e.g., Amazon Nova, Anthropic Claude, Meta Llama, Mistral) through **one API**. Adds Knowledge Bases (RAG), Agents, and Guardrails. Your data isn't used to train the base models |
| **Amazon SageMaker JumpStart** | ML hub inside SageMaker AI with **pre-built models** (vision, language, tabular) you can fine-tune and deploy quickly |
| **Amazon Q Business** | GenAI assistant that answers questions using your **company's data** through secure connections to business systems |
| **Amazon Q Developer** | GenAI assistant for **developers**: code suggestions, whole functions, AWS troubleshooting |

> ⚠️ **2026 update:** AWS renames AI products often. QuickSight became part of **Amazon Quick** (launched as Quick Suite in Oct 2025), which AWS positions as the next step beyond Amazon Q Business. Amazon Q Developer still exists, with **Kiro** as AWS's newer AI coding tool. Learn the course names for the exam, but know the current ones for customer conversations.

### Introduction to Data Analytics

- **Data analytics** = turning raw data into **insights** that drive decisions (often from **historical** data)
- **Four types of analytics:**

| Type | Question it answers | Example |
|---|---|---|
| **Descriptive** | What happened? | Last quarter's sales by region |
| **Diagnostic** | Why did it happen? | Sales dropped because of a stock-out |
| **Predictive** | What will happen? | Forecast next quarter's demand |
| **Prescriptive** | What should we do? | Recommend how much inventory to order |

- **Data types:** **structured** (tables), **semi-structured** (JSON, logs), **unstructured** (images, PDFs, audio)
- **Batch** = process data in chunks on a schedule (nightly sales report) | **Streaming** = process data continuously as it arrives (fraud alerts in seconds)
- **Data lake** = store **all** data in raw form, cheaply (usually S3) | **Data warehouse** = structured, cleaned data optimized for fast SQL analytics (e.g., Redshift)

### Data Pipelines on AWS
**Data pipeline** = the automated path data takes from its source to insights. **ETL** = Extract, Transform, Load.

| Stage | What happens | AWS services |
|---|---|---|
| **1. Ingestion** | Collect data and move it into storage | **Amazon Kinesis Data Streams** (real-time streams), **Amazon Data Firehose** (loads streaming data into storage in near real time) |
| **2. Storage** | Centralize data from many sources | **Amazon S3** (data lake), **Amazon Redshift** (data warehouse) |
| **3. Cataloging** | Organize and find data: what exists, where, what schema | **AWS Glue Data Catalog** |
| **4. Processing** | Clean, transform, and enrich data | **AWS Glue** (serverless ETL), **Amazon EMR** (managed Spark, Hadoop, Hive for big data) |
| **5. Analysis & visualization** | Query data and build insights | **Amazon Athena** (SQL directly on S3, pay per query), **Amazon Redshift** (fast SQL on large datasets), **Amazon QuickSight** (dashboards and BI), **Amazon OpenSearch Service** (search and real-time log analytics) |

### Data Analytics and AI/ML

- Analytics and AI/ML **work together**: analytics pipelines produce the **clean, organized data** ML models need, and model predictions feed back into dashboards and decisions
- **Garbage in = garbage out:** the quality of AI depends on the quality and governance of the data
- Example: a retailer uses analytics to clean years of sales data (descriptive), trains an ML model to forecast demand (predictive), and surfaces recommendations in a dashboard (prescriptive)
