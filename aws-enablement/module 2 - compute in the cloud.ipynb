## MODULE 2 INTRODUCTION

### Introduction to Amazon EC2
- **Amazon EC2 (Elastic Compute Cloud)** = secure, resizable **virtual servers** ("instances") in the AWS Cloud
- **Compute** = the processing power (CPU, memory) that runs your applications
- **On-prem vs EC2:** on-prem means buying servers, waiting weeks for delivery, and paying even when they sit idle. EC2 = launch in minutes, stop paying when you terminate
- **Multitenancy:** many instances share one physical host. The **hypervisor** (managed by AWS) keeps each instance isolated from the others
- **You control:** the OS (Linux/Windows), software, networking config. That means **you** patch the OS (Shared Responsibility)
- EC2 is **IaaS** (Infrastructure as a Service)

## COMPUTE IN THE CLOUD

### Amazon EC2 Instance Types
Instance types are grouped into **families**, each optimized for a different workload.

| Family | Optimized for | Example use |
|---|---|---|
| **General purpose** | Balance of compute, memory, networking | Web/app servers, code repositories, gaming servers |
| **Compute optimized** | High-performance processors | Batch processing, scientific modeling, high-performance web servers |
| **Memory optimized** | Processing large datasets in memory | In-memory databases, real-time big data analytics |
| **Accelerated computing** | Hardware accelerators (GPUs, custom chips) | Machine learning, graphics, floating-point calculations |
| **Storage optimized** | High read/write to large local datasets | Data warehousing, distributed file systems, high-frequency OLTP |

> 💡 Instance names follow a pattern, e.g. `m7g.large`: **m** = family (general purpose), **7** = generation, **g** = processor option (Graviton), **large** = size.

### How to Provision AWS Resources
Every action in AWS is an **API call**. Three main ways to make those calls:

| Method | What it is | Best for |
|---|---|---|
| **AWS Management Console** | Web-based point-and-click interface | Learning, one-off tasks, visual exploration |
| **AWS CLI** (Command Line Interface) | Commands typed in a terminal; scriptable | Automating repetitive tasks |
| **AWS SDKs** (Software Development Kits) | Code libraries (Python/boto3, Java, JavaScript, etc.) | Calling AWS from inside your applications |

> 💡 Infrastructure as Code (CloudFormation) is covered later in the course (Module 4).

### Demo: Launching an Amazon EC2 Instance
Steps in the console:
1. **Name** the instance (and add tags)
2. Choose an **AMI** (Amazon Machine Image) = the template: OS + preinstalled software
3. Choose an **instance type** (size/family)
4. Choose a **key pair** (used to securely connect/log in)
5. Configure **network settings** + **security group** (virtual firewall controlling allowed traffic)
6. Configure **storage** (EBS volume)
7. **Launch** → then **connect** to the instance

### Amazon EC2 Pricing
| Option | How it works | Best for |
|---|---|---|
| **On-Demand** | Pay by the second/hour, no commitment | Short-term, irregular, unpredictable workloads; testing |
| **Savings Plans** | Commit to a consistent $/hour of usage for 1 or 3 years | Steady usage, flexible across instance types (up to ~72% off) |
| **Reserved Instances** | Commit to a specific instance config for 1 or 3 years (Standard or Convertible) | Steady-state, predictable workloads |
| **Spot Instances** | Use spare AWS capacity; AWS can reclaim with a **2-minute warning** | Flexible, interruptible jobs (batch, background processing); up to ~90% off |
| **Dedicated Hosts** | A physical server fully dedicated to you | Compliance needs, existing server-bound software licenses (most expensive) |

> 💡 Payment options for commitments: All Upfront (biggest discount), Partial Upfront, No Upfront.

## AUTO SCALING AND LOAD BALANCING

### Scaling Amazon EC2
- **Scalability** = start with only the resources you need and adjust automatically to demand
- **Amazon EC2 Auto Scaling** = automatically adds or removes instances
  - **Dynamic scaling** → responds to changing demand
  - **Predictive scaling** → schedules capacity based on predicted demand
- **Auto Scaling group** settings:
  - **Minimum capacity** = instances that always run
  - **Desired capacity** = the target number (defaults to minimum)
  - **Maximum capacity** = upper limit (controls cost)
- **Scale out/in** = add/remove instances (horizontal) | **Scale up/down** = bigger/smaller instance (vertical)

### Directing Traffic with Elastic Load Balancing
- **Elastic Load Balancing (ELB)** = automatically distributes incoming traffic across multiple resources (e.g., EC2 instances)
- Single point of contact for users; no instance gets overwhelmed
- Managed by AWS and **scales automatically**; works across AZs
- Routes traffic only to **healthy** instances (health checks)
- **Auto Scaling + ELB together** = scale with demand while keeping traffic evenly spread

**Analogy:** Auto Scaling decides **how many** cashiers are working; the load balancer is the **host** pointing each customer to an open register.

### Messaging and Queuing
- **Tightly coupled** = components talk directly; if one fails, the failure cascades ❌
- **Loosely coupled** = a buffer sits between components; one failure doesn't take down the rest ✅
- **Monolithic** app vs **microservices** (small independent services that communicate)

| Service | Model | What it does |
|---|---|---|
| **Amazon SQS** (Simple Queue Service) | **Queue** (pull) | Sends, stores, and receives messages between components; messages wait in the queue until processed |
| **Amazon SNS** (Simple Notification Service) | **Publish/subscribe** (push) | Publishes a message to a **topic**; all subscribers (email, SMS, Lambda, SQS) receive it |
| **Amazon EventBridge** | **Event bus** | Routes events between AWS services, your apps, and SaaS apps based on rules |

**Memory trick:** SQS = a **to-do inbox** someone works through | SNS = a **broadcast** to everyone subscribed

## MODULE 2 CONCLUSION

### Module Summary
- EC2 = on-demand virtual servers; you manage the OS
- Pick the **instance family** that matches the workload
- Provision via **Console, CLI, or SDKs** (all API calls)
- Pick the **pricing option** that matches the usage pattern
- **Auto Scaling** matches capacity to demand; **ELB** spreads traffic
- **SQS/SNS/EventBridge** decouple applications

## 🎯 Databricks SA Connection
- **Classic Databricks clusters run on EC2** in the customer's AWS account. Choosing a cluster node type = choosing an EC2 instance family
  - Memory optimized → large joins/aggregations | Compute optimized → heavy transformations | Accelerated (GPU) → deep learning | Storage optimized → heavy disk caching
- **Spot is a top cost lever:** Spot worker nodes + On-Demand driver node. Lost Spot workers get replaced without killing the job
- **Savings Plans / RIs:** classic compute uses the customer's EC2, so their existing AWS commitments can apply
- **Auto Scaling ↔ Databricks autoscaling clusters** (same idea: min/max workers)
- **Serverless Databricks** removes instance choices entirely (Databricks manages the compute)
- **SQS/SNS in data pipelines:** S3 can publish file-arrival events to SNS/SQS, which is how Databricks **Auto Loader** (file notification mode) efficiently discovers new files
