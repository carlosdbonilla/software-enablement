## MODULE 11 INTRODUCTION

## PRICING AND SUPPORT

### AWS Pricing Concepts
**Three pricing principles:**

| Principle | What it means | Example |
|---|---|---|
| **Pay as you go** | Pay only for what you use, no long-term contract required | On-Demand EC2 billed per second |
| **Save when you commit** | Commit to steady usage for **1 or 3 years** → big discounts | Savings Plans, Reserved Instances |
| **Pay less by using more** | **Tiered (volume) pricing**: unit price drops as usage grows | S3 per-GB price falls at higher storage tiers |

**Three main cost drivers:**

| Driver | How it's charged |
|---|---|
| **Compute** | By time (per second/hour) while resources run |
| **Storage** | By amount stored (GB-month) and storage class |
| **Data transfer** | **Inbound to AWS is free**. **Outbound to the internet** is charged per GB (cheaper at volume). Transfer **between AZs or Regions** also costs money |

> 💡 Data transfer is the "hidden" cost: moving data **out** of AWS or **across Regions** adds up fast.

**AWS Free Tier:**

- **Always free:** certain services have a free allowance that never expires (e.g., a monthly amount of Lambda requests and DynamoDB storage)
- **Free trials:** short-term free use of specific services
- ⚠️ **2026 update:** since **July 15, 2025**, new accounts get **up to $200 in credits** and can choose a **free plan** (up to 6 months, or until credits run out) instead of the old "12 months free" model. Older material may still describe 12-month free offers

### AWS Pricing and Billing Services

| Tool | Use it to… | When |
|---|---|---|
| **AWS Pricing Calculator** | **Estimate** costs of a planned architecture (instance types, storage, data transfer) | **Before** you build |
| **AWS Billing and Cost Management** (dashboard) | See current charges, usage, forecasts, invoices, payments | **Now** |
| **AWS Cost Explorer** | **Visualize and analyze** spend over time with graphs and filters (by service, account, tag); forecasts; Savings Plan/RI recommendations | **Past + trend** |
| **AWS Budgets** | Set custom **cost, usage, or Savings Plan/RI** budgets and get **alerts** when you exceed (or are forecast to exceed) them | **Ongoing control** |
| **AWS Organizations: consolidated billing** | **One bill** for all accounts; usage combined for **volume discounts**; commitment discounts can be shared across accounts | **Multi-account** |
| **Cost allocation tags** | Label resources (team, project, cost center) to break down spend | **Chargeback/showback** |
| **AWS Cost and Usage Report (CUR)** | Most **detailed** line-item billing data for deep analysis | **Finance/FinOps analysis** |

> 💡 Memory trick: **Calculator = predict**, **Cost Explorer = analyze the past**, **Budgets = alert me**.

### AWS Support Plans
**⚠️ AWS is restructuring support.** The course and current exam questions use the **classic** plans, so learn those first, then know the change for real customer conversations.

**Classic plans (learn these for the exam):**

| Plan | Best for | Fastest response (critical case) | Key features |
|---|---|---|---|
| **Basic** (free) | All customers | No technical cases | 24/7 customer service for billing/account, docs, whitepapers, re:Post, AWS Health Dashboard, **core** Trusted Advisor checks |
| **Developer** | Testing/experimenting | **< 12 business hours** (system impaired) | Business-hours **email** access to support; general guidance < 24 business hours |
| **Business** | **Production** workloads (minimum recommended) | **< 1 hour** (production system down) | **24/7 phone, email, chat**; **full** Trusted Advisor; AWS Health API; third-party software support |
| **Enterprise On-Ramp** | Production + business-critical | **< 30 minutes** (business-critical system down) | **Pool of TAMs**, Concierge support team |
| **Enterprise** | Business- and **mission-critical** | **< 15 minutes** (business-critical system down) | **Designated TAM**, Concierge, Well-Architected reviews, Infrastructure Event Management |

**Technical Account Manager (TAM)** = your **named AWS advisor**: proactive guidance on architecture, operations, and cost, and your connection to AWS experts and programs.

**New structure (2026 → 2027):**

| New plan | Replaces / notes |
|---|---|
| **Basic** | Unchanged, free |
| **Business Support+** | Replaces **Developer and Business**; AI-powered troubleshooting plus 24/7 access to AWS experts; starts around $29/month per account |
| **Enterprise Support** | Absorbs **Enterprise On-Ramp** (customers auto-upgraded during 2026); starts around $5,000/month |
| **Unified Operations** | New top tier for the most critical operations; starts around $50,000/month |

- **Developer, Business, and Enterprise On-Ramp are discontinued January 1, 2027** (they remain in AWS GovCloud (US))

**Exam shortcuts:** "designated TAM" → **Enterprise** | "pool of TAMs" → **Enterprise On-Ramp** | "24/7 phone at lowest cost" → **Business** | "business-hours email, cheapest technical support" → **Developer** | "billing questions only" → **Basic**

### AWS Marketplace and AWS Partners
**AWS Marketplace** = digital catalog of **thousands of third-party software products** that run on AWS.

- Find, **try (free trials)**, buy, and deploy software; see pricing options and customer reviews
- Categories include **SaaS**, **ML/AI** (pre-built models), **data and analytics**, security, DevOps, and more
- **Billed through your AWS bill**; flexible pricing (pay-as-you-go, annual, private offers)

**AWS Partner Network (APN)** = global community of companies that build solutions and services on AWS.

- **Services/consulting partners:** help you design, migrate, and manage workloads
- **Software/technology partners:** build products that run on or integrate with AWS
- Partners earn **competencies** (validated expertise, e.g., Data & Analytics, Migration); find them via the Partner Solutions Finder
- Partner benefits: funding programs, events, training and certification

**Other help resources:** AWS Professional Services (AWS's own consultants), AWS re:Post (community Q&A), AWS Knowledge Center, documentation, and AWS Trust & Safety (report abuse of AWS resources).

### Cost Optimization

| Area | Strategy | AWS tool |
|---|---|---|
| **Compute** | **Rightsize** instances to match actual need | **AWS Compute Optimizer** |
| | Use **Spot** for interruptible work, **Savings Plans/RIs** for steady work | Cost Explorer recommendations |
| | **Auto Scaling** removes excess capacity when demand drops; turn off idle dev/test resources | EC2 Auto Scaling |
| **Databases** | Rightsize RDS; use **storage autoscaling**; add **read replicas** instead of a bigger instance; **cache** hot data | RDS, **ElastiCache** |
| **Storage** | Pick the right **S3 storage class**; use **Intelligent-Tiering** for unknown access; **lifecycle policies**; **Glacier Deep Archive** for rarely accessed data | S3 |
| **Networking** | Use **VPC endpoints** to reach S3 privately and cut data transfer/NAT costs | VPC endpoints |
| **Visibility** | Track, alert, and find waste | **Cost Explorer**, **Budgets**, **Trusted Advisor**, Cost Anomaly Detection |

> 💡 Cost optimization is continuous, not one-time: **measure → find waste → act → repeat.** It's also a pillar of the Well-Architected Framework (Module 13).

## 🎯 Databricks SA Connection

- **Databricks is sold through AWS Marketplace.** Big deal in enterprise sales: buying Databricks through Marketplace can **count toward the customer's AWS spending commitment** (their enterprise discount agreement) and simplifies procurement because it's on the AWS bill
- **Databricks is an AWS Partner.** Joint selling with AWS account teams is common. Treat the AWS rep as an ally, not a competitor
- **Two bills on classic compute:** customers pay **Databricks (DBUs)** for the platform **plus AWS** for the underlying EC2, EBS, and networking. **Serverless** bundles the compute into Databricks pricing. Be ready to explain this clearly. It's one of the most common customer questions
- **TCO conversations** use the same mindset as the Pricing Calculator: estimate workload costs and compare to the current stack (e.g., EMR + Glue + Redshift + separate ML tools)
- **Cost levers in Databricks:** serverless, autoscaling, **auto-termination** of idle clusters, Spot workers, Photon, compute policies, and right-sized warehouses
- **Data transfer fees** are why the workspace should be in the **same Region** as the data
- **Cost visibility:** Databricks **billing system tables** and account-level **budgets** give a Cost Explorer-style view of platform spend
- **Discovery question:** "Do you have an AWS commit you're trying to burn down?" (this can make Marketplace purchasing very attractive)
